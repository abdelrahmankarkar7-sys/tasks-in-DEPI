```python
class TodoList:
    def __init__(self):
        self.tasks = []

    def add_task(self, task):
        self.tasks.append({"task": task, "done": False})

    def complete_task(self, task):
        for item in self.tasks:
            if item["task"] == task:
                item["done"] = True


todo = TodoList()

todo.add_task("Study Python")
todo.add_task("Complete homework")

todo.complete_task("Study Python")

print(todo.tasks)
```

```python
class Stack:
    def __init__(self):
        self.items = []

    def push(self, item):
        self.items.append(item)

    def pop(self):
        if self.items:
            return self.items.pop()
        return None


stack = Stack()

stack.push(1)
stack.push(2)
stack.push(3)

popped = stack.pop()

print("Popped:", popped)
print("Remaining stack:", stack.items)
```

```python
class Queue:
    def __init__(self):
        self.items = []

    def enqueue(self, item):
        self.items.append(item)

    def dequeue(self):
        if self.items:
            return self.items.pop(0)
        return None


queue = Queue()

queue.enqueue("a")
queue.enqueue("b")
queue.enqueue("c")

dequeued = queue.dequeue()

print("Dequeued:", dequeued)
print("Remaining queue:", queue.items)
```

```python
class Movie:
    def __init__(self, title, rating):
        self.title = title
        self.rating = rating


class MovieCollection:
    def __init__(self):
        self.movies = []

    def add(self, movie):
        self.movies.append(movie)

    def average_rating(self):
        if not self.movies:
            return 0

        total = sum(movie.rating for movie in self.movies)
        return total / len(self.movies)


collection = MovieCollection()

collection.add(Movie("Movie 1", 8))
collection.add(Movie("Movie 2", 6))
collection.add(Movie("Movie 3", 10))

print(collection.average_rating())
```

```python
class Inventory2:
    def __init__(self):
        self.stock = {}

    def add_stock(self, item, qty):
        if item in self.stock:
            self.stock[item] += qty
        else:
            self.stock[item] = qty

    def remove_stock(self, item, qty):
        if item in self.stock and self.stock[item] >= qty:
            self.stock[item] -= qty
            return True
        return False


inventory = Inventory2()

inventory.add_stock("pen", 50)
inventory.add_stock("pen", 20)

print(inventory.remove_stock("pen", 30))
print(inventory.remove_stock("pen", 100))
print(inventory.stock)
```

```python
class BMI:
    def __init__(self, weight, height):
        self.weight = weight
        self.height = height

    def calculate(self):
        return self.weight / (self.height ** 2)

    def category(self):
        bmi = self.calculate()

        if bmi < 18.5:
            return "Underweight"
        elif bmi < 25:
            return "Normal"
        elif bmi < 30:
            return "Overweight"
        else:
            return "Obese"


bmi = BMI(70, 1.75)

print(bmi.calculate())
print(bmi.category())
```

```python
class Game:
    def __init__(self):
        self.score = 0
        self.high_score = 0

    def add_points(self, pts):
        self.score += pts

        if self.score > self.high_score:
            self.high_score = self.score

    def reset(self):
        self.score = 0


game = Game()

game.add_points(50)
game.add_points(30)
game.reset()
game.add_points(20)

print("Score:", game.score)
print("High score:", game.high_score)
```

```python
class Employee3:
    def __init__(self, name, hours, rate):
        self.name = name
        self.hours = hours
        self.rate = rate

    def calculate_pay(self):
        if self.hours <= 40:
            return self.hours * self.rate
        else:
            normal_pay = 40 * self.rate
            overtime_hours = self.hours - 40
            overtime_pay = overtime_hours * self.rate * 1.5
            return normal_pay + overtime_pay


employee1 = Employee3("Ahmed", 45, 20)
employee2 = Employee3("Mohamed", 35, 25)

print(employee1.calculate_pay())
print(employee2.calculate_pay())
```

```python
class Node2:
    def __init__(self, value):
        self.value = value
        self.next = None


class LinkedListSimple:
    def __init__(self):
        self.head = None

    def add(self, value):
        new_node = Node2(value)

        if self.head is None:
            self.head = new_node
            return

        current = self.head

        while current.next is not None:
            current = current.next

        current.next = new_node

    def to_list(self):
        result = []
        current = self.head

        while current is not None:
            result.append(current.value)
            current = current.next

        return result


linked_list = LinkedListSimple()

linked_list.add(10)
linked_list.add(20)
linked_list.add(30)

print(linked_list.to_list())
```
