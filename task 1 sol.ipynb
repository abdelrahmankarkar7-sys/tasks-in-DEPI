1. List Operations

```python
numbers = [1, 2, 3, 4, 5]
numbers.append(6)
numbers[1] = 10
print(numbers)
```

2. List Manipulation

```python
my_list = ['apple', 'banana', 'cherry', 'date']
print(my_list[0])
```

```python
my_list.pop()
my_list.insert(2, 'grape')
print(my_list)
```

3. Set Operations

```python
numbers = {1, 2, 3, 4}
numbers.add(5)
numbers.add(3)
numbers.remove(2)
print(numbers)
```

4. Set Manipulation

```python
my_set = {'red', 'green', 'blue'}
print('green' in my_set)
```

```python
my_set.add('yellow')
my_set.remove('red')
print(my_set)
```

5. Tuple Operations

```python
numbers = (1, 2, 3, 4, 5)
print(numbers[2])
print(numbers)
```

6. Tuple Manipulation

```python
my_tuple = ('apple', 'banana', 'cherry')
print(my_tuple[-1])
print('banana' in my_tuple)
print(my_tuple)
```

7. Dictionary Operations

```python
person = { 'name': 'John', 'age': 25,  'city': 'New York'}
print(person['name'])
```

```python
person['age'] = 26
person['job'] = 'Engineer'
print(person)
```

8. Dictionary Manipulation

```python
my_dict = {'brand': 'Toyota','model': 'Corolla','year': 2019}
print(my_dict['model'])
```

```python
my_dict.pop('year')
my_dict['color'] = 'blue'
print(my_dict)
```
